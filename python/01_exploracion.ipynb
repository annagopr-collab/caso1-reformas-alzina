```python
import pandas as pd
from pathlib import Path
```

```python
carpeta = Path("../datos_brutos")
```

```python
for fichero in sorted(carpeta.iterdir()):
    print("=" * 80)
    print(fichero.name)
    try:
        if fichero.suffix == ".xlsx":
            df = pd.read_excel(fichero)
        else:
            df = pd.read_csv(fichero, sep=None, engine="python")
    except UnicodeDecodeError:
        df = pd.read_csv(fichero, sep=None, engine="python", encoding="latin-1")
        print("(este fichero no está en UTF-8: lo he leído como latin-1)")
    print(f"{df.shape[0]} filas x {df.shape[1]} columnas")
    print(df.head(3).to_string())
```
