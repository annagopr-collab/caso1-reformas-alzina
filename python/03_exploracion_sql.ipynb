## Facturas antiguas
### 1. ¿Cuántas facturas hay?

```python
sql("SELECT COUNT(*) AS facturas FROM bruto_facturas_antiguas")
```

### 2. ¿Hay algún número de factura repetido?

```python
sql("""
SELECT n_factura, COUNT(*) AS veces
FROM bruto_facturas_antiguas
GROUP BY n_factura
HAVING COUNT(*) > 1
""")
```

### 3. ¿Qué valores tiene el IVA y cuántas facturas hay de cada uno?

```python
sql("""
SELECT pct_iva, COUNT(*) AS facturas
FROM bruto_facturas_antiguas
GROUP BY pct_iva
""")
```

### 4. ¿Ves el mismo cliente escrito de varias formas?

```python
pd.set_option("display.max_rows", 200)
sql("""
SELECT cliente, COUNT(*) AS facturas
FROM bruto_facturas_antiguas
GROUP BY cliente
ORDER BY cliente
""")
```

### 5. ¿Hay facturas sin NIF?

```python
sql("""
SELECT COUNT(*) AS sin_nif
FROM bruto_facturas_antiguas
WHERE nif_cif IS NULL OR nif_cif = ''
""")
```
